```python
from graphrag.config.load_config import load_config
from pathlib import Path
from graphrag.api.index import build_index
import os
import shutil
```

```python
conf = load_config(Path("./"))
gr_conf = await build_index(config=conf)
```

```python
# only list the directories in ./
list_dirs = [d for d in os.listdir("./") if os.path.isdir(os.path.join("./", d))]

# Delete the same dirs from ../cache/kbgraph
for d in list_dirs:
    if os.path.exists(os.path.join("../cache/kbgraph", d)):
        shutil.rmtree(os.path.join("../cache/kbgraph", d))

# Copy the dirs to ../cache/kbgraph
for d in list_dirs:
    shutil.copytree(os.path.join("./", d), os.path.join("../cache/kbgraph", d))
```
